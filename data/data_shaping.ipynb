```python
import pandas as pd
```

```python
data = pd.read_csv('sahko_data30.5.csv')
data.head()
data.tail()
```

```python
data2 = pd.read_csv('combined3.7.csv')
#data2.drop(columns=['startDate'], inplace=True)
data2.head()
data2.tail()
```

```python
# Combine previously loaded dataframes `data` and `data2` and save to CSV
import pandas as pd
from pathlib import Path

# Concatenate row-wise and remove exact duplicate rows
combined = pd.concat([data, data2], ignore_index=True)
combined.sort_values(by='time', inplace=True)
combined = combined.drop_duplicates()

# Write to combined.csv
out_path = Path("combined3.8.csv")
out_path.parent.mkdir(parents=True, exist_ok=True)
combined.to_csv(out_path, index=False)
print(f"Wrote {len(combined)} rows to {out_path}")
```
